```python
# import folium
# import pandas as pd
# from IPython.display import display

# # Thiết lập bản đồ Folium
# lat = 20.97
# long = 105.8
# zoom = 12
# gmap2 = folium.Map(location=(lat, long), zoom_start=zoom)

# # Hàm thêm marker vào bản đồ
# def _addMarker(gmap):
#     for i, row in dataset.iterrows():
#         marker = folium.Marker(location=(row['DELIVERY_WARD_LATITUDE'], row['DELIVERY_WARD_LONGITUDE']),
#                                popup=row['TenTruong'])
#         marker.add_to(gmap)

# # Gọi hàm để thêm marker vào bản đồ
# _addMarker(gmap2)

# # Hiển thị bản đồ trực tiếp trong Colab
# display(gmap2)  # Dùng trực tiếp đối tượng Folium map
# day la ban thu
# import folium
# import pandas as pd
# from IPython.display import display
# import osmnx as ox
# import networkx as nx

# # Tọa độ của trường Đại học Điện lực
# daihoc_dienluc_lat = 21.046548190369
# daihoc_dienluc_long = 105.784986510929

# # Tạo bản đồ Folium với tọa độ trung tâm
# lat = 21.04
# long = 105.8
# zoom = 13
# gmap2 = folium.Map(location=(lat, long), zoom_start=zoom)

# # Tải mạng lưới đường trong phạm vi xung quanh khu vực
# G = ox.graph_from_point((daihoc_dienluc_lat, daihoc_dienluc_long), dist=5000, network_type='drive')

# # Đọc dữ liệu từ tệp Excel
# dataset = pd.read_excel("/content/drive/MyDrive/ML/data/listtruongcaugiay.xlsx", header=0)
# dataset.columns = ['DELIVERY_WARD_LATITUDE', 'DELIVERY_WARD_LONGITUDE', 'TenTruong', 'STT']

# # Hàm để tìm đường đi ngắn nhất và vẽ đường bộ trên bản đồ
# def add_route(gmap, G, start_lat, start_lon, end_lat, end_lon):
#     # Tìm nút gần nhất đến điểm bắt đầu và kết thúc
#     start_node = ox.distance.nearest_nodes(G, start_lon, start_lat)
#     end_node = ox.distance.nearest_nodes(G, end_lon, end_lat)

#     # Tính toán đường đi ngắn nhất theo khoảng cách
#     route = nx.shortest_path(G, start_node, end_node, weight='length')

#     # Lấy tọa độ của tuyến đường
#     route_coords = [(G.nodes[node]['y'], G.nodes[node]['x']) for node in route]

#     # Vẽ đường đi ngắn nhất lên bản đồ
#     folium.PolyLine(
#         locations=route_coords,
#         color="green",
#         weight=2.5,
#         opacity=0.7
#     ).add_to(gmap)

# # Thêm các marker và đường đi ngắn nhất từ Đại học Điện lực đến các điểm khác
# for i, row in dataset.iterrows():
#     end_lat = row['DELIVERY_WARD_LATITUDE']
#     end_lon = row['DELIVERY_WARD_LONGITUDE']

#     # Thêm marker cho từng trường
#     folium.Marker(
#         location=(end_lat, end_lon),
#         popup=row['TenTruong']
#     ).add_to(gmap2)

#     # Vẽ đường đi ngắn nhất
#     add_route(gmap2, G, daihoc_dienluc_lat, daihoc_dienluc_long, end_lat, end_lon)

# # Hiển thị bản đồ trong Colab
# display(gmap2)
import folium
import pandas as pd
from IPython.display import display
from math import radians, sin, cos, sqrt, atan2

# Tọa độ trường Đại học Điện lực
daihoc_dienluc_lat = 21.046548190369
daihoc_dienluc_long = 105.784986510929

# Thiết lập bản đồ Folium
lat = 20.97
long = 105.8
zoom = 12
gmap2 = folium.Map(location=(lat, long), zoom_start=zoom)

# Hàm tính khoảng cách giữa hai tọa độ
def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Bán kính Trái Đất tính theo km
    dlat = radians(lat2 - lat1)
    dlon = radians(lon2 - lon1)
    a = sin(dlat / 2) ** 2 + cos(radians(lat1)) * cos(radians(lat2)) * sin(dlon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = R * c
    return distance

# Hàm thêm marker vào bản đồ, tính khoảng cách và vẽ đường nối
def _addMarkerAndLine(gmap, dataset):
    for i, row in dataset.iterrows():
        location = (row['DELIVERY_WARD_LATITUDE'], row['DELIVERY_WARD_LONGITUDE'])
        distance = haversine(daihoc_dienluc_lat, daihoc_dienluc_long, location[0], location[1])

        # Thêm marker với khoảng cách từ trường Đại học Điện lực
        marker = folium.Marker(
            location=location,
            popup=f"{row['TenTruong']} - Khoảng cách: {distance:.2f} km"
        )
        marker.add_to(gmap)

        # Vẽ đường nối từ Đại học Điện lực đến điểm hiện tại
        folium.PolyLine(
            locations=[(daihoc_dienluc_lat, daihoc_dienluc_long), location],
            color="blue",
            weight=2.5,
            opacity=0.7
        ).add_to(gmap)

# Đọc dữ liệu từ tệp Excel
dataset = pd.read_excel("/content/drive/MyDrive/ML/data/listtruongcaugiay.xlsx", header=0)
dataset.columns = ['DELIVERY_WARD_LATITUDE', 'DELIVERY_WARD_LONGITUDE', 'TenTruong', 'STT']

# Gọi hàm để thêm marker và vẽ đường nối
_addMarkerAndLine(gmap2, dataset)

# Hiển thị bản đồ trực tiếp trong Colab
display(gmap2)
```

```python
import folium
import pandas as pd
import osmnx as ox
import networkx as nx

# Tọa độ của trường Đại học Điện lực
daihoc_dienluc_lat = 21.046548190369
daihoc_dienluc_long = 105.784986510929

# Tạo bản đồ Folium với tọa độ trung tâm
lat = 21.04
long = 105.8
zoom = 13
gmap2 = folium.Map(location=(lat, long), zoom_start=zoom)

# Tải mạng lưới đường trong phạm vi xung quanh khu vực Đại học Điện lực
G = ox.graph_from_point((daihoc_dienluc_lat, daihoc_dienluc_long), dist=5000, network_type='drive')

# Đọc dữ liệu từ tệp Excel
dataset = pd.read_excel("/content/drive/MyDrive/ML/data/listtruongcaugiay.xlsx", header=0)
dataset.columns = ['DELIVERY_WARD_LATITUDE', 'DELIVERY_WARD_LONGITUDE', 'TenTruong', 'STT']

# Hàm để tìm đường đi ngắn nhất và vẽ đường bộ trên bản đồ
def add_route(gmap, G, start_lat, start_lon, end_lat, end_lon):
    # Tìm nút gần nhất đến điểm bắt đầu và kết thúc
    start_node = ox.distance.nearest_nodes(G, start_lon, start_lat)
    end_node = ox.distance.nearest_nodes(G, end_lon, end_lat)

    # Tính toán đường đi ngắn nhất theo khoảng cách
    route = nx.shortest_path(G, start_node, end_node, weight='length')

    # Lấy tọa độ của tuyến đường
    route_coords = [(G.nodes[node]['y'], G.nodes[node]['x']) for node in route]

    # Vẽ đường đi ngắn nhất lên bản đồ
    folium.PolyLine(
        locations=route_coords,
        color="green",
        weight=2.5,
        opacity=0.7
    ).add_to(gmap)

# Thêm các marker và đường đi ngắn nhất từ Đại học Điện lực đến các điểm khác
for i, row in dataset.iterrows():
    end_lat = row['DELIVERY_WARD_LATITUDE']
    end_lon = row['DELIVERY_WARD_LONGITUDE']

    # Thêm marker cho từng trường
    folium.Marker(
        location=(end_lat, end_lon),
        popup=row['TenTruong']
    ).add_to(gmap2)

    # Vẽ đường đi ngắn nhất từ Đại học Điện lực đến trường này
    add_route(gmap2, G, daihoc_dienluc_lat, daihoc_dienluc_long, end_lat, end_lon)

# Hiển thị bản đồ
gmap2
```
